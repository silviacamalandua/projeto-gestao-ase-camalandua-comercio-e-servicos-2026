```python
#Kpis

#primeira KPI - quantidade tarefas concluidas durante os 5 meses
quantidade_concluidas = (df_meses["status"] == "Finalizada").sum()
print(quantidade_concluidas)

#opcional - tarefas pendentes, aguardando validação e em andamento
quantidade_em_andamento = (df_meses["status"] == "Em andamento").sum()
print(quantidade_em_andamento)
quantidade_validacao = (df_meses["status"] == "Aguardando validação").sum()
print(quantidade_validacao)
quantidade_pendente = (df_meses["status"] == "Pendente").sum()
print(quantidade_pendente)

#segunda KPI - percentual de tarefas concluidas 
percentual = (quantidade_concluidas/len(df_meses))*100
print(percentual)

#terceiro KPI - Tempo Médio de Execução
#Tanto para as tarefas concluídas, quanto para as que estão em aguardando validação
tarefas_concluidas = df_meses [df_meses ["status"].isin(["Finalizada", "Aguardando validação"])]
tempo_medio = tarefas_concluidas["tempo_execucao"].mean()
print(tempo_medio)

#quarta KPI - quantidade de tarefas por usuário
quantidade_tarefas = (df_meses.groupby(["id_usuario_responsavel","nome"]).size())
print(quantidade_tarefas)

#opcional - quantidade de tarefas finalizadas por usuário
quantidade_tarefas_finalizada = (df_meses[df_meses["status"] == "Finalizada"].groupby(["id_usuario_responsavel","nome"]).size())
print(quantidade_tarefas_finalizada)

```

```python
#Estastística Descritiva

#Média
tempo_medio = tarefas_concluidas["tempo_execucao"].mean()
print(tempo_medio)

#Mediana 
mediana = tarefas_concluidas["tempo_execucao"].median()
print(mediana)

#Desvio Padrão
desvio_padrao = tarefas_concluidas["tempo_execucao"].std()
print(desvio_padrao)
```

```python
#Distribuição e Possíveis Outliers

Q1 = tarefas_concluidas["tempo_execucao"].quantile(0.25)
Q3 = tarefas_concluidas["tempo_execucao"].quantile(0.75)
IQR = Q3 - Q1

print("Q1 : ",Q1)
print("Q3 : ",Q3)
print("IQR: ",IQR)

limite_inferior = Q1 - 1.5*IQR
limite_superior = Q3 + 1.5*IQR
print("Limite inferior = ",limite_inferior)
print("Limite superior = ",limite_superior)

outliers = tarefas_concluidas[
    (tarefas_concluidas["tempo_execucao"]< limite_inferior)|(tarefas_concluidas["tempo_execucao"] > limite_superior)
]
print(outliers[["id_tarefa","nome","tempo_execucao"]])

outliers[["id_tarefa","nome","tipo_atividade","tempo_execucao"]]
#Tempo médio de execução por atividade
TM_atividade = (tarefas_concluidas.groupby("tipo_atividade")["tempo_execucao"].mean())
print(TM_atividade)
```
